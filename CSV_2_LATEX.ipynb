```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
```

```python
# Load CSV
path = "./outputs/output_german_credit_standard.csv"
df = pd.read_csv(path)
df = df.drop(['Dataset', 'Variant'], axis=1)

df['Noise Level'] = df['Noise Level'].str.replace(r'noise_', '', regex=True)
df['accuracy'] = df['accuracy'].apply(lambda x: f"{x:.2g}" if pd.notna(x) else x)
df['precision'] = df['precision'].apply(lambda x: f"{x:.2g}" if pd.notna(x) else x)

df = df.fillna("NA")


latex_table = df.to_latex(index=False)

# Wrap in a complete LaTeX table 
wrapped_latex_table = r"""
\begin{table}[!htbp]
    \centering
    \scalebox{0.37}{
""" + latex_table + r"""
    }
    \caption{}
    \label{}
\end{table}
"""

print(wrapped_latex_table)
```
